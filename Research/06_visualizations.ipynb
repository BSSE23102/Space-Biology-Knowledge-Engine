# Space Biology Knowledge Engine - Visualizations

This notebook creates comprehensive visualizations of the space biology research landscape.

## Objectives:
1. Create t-SNE visualizations of embeddings
2. Generate word clouds and topic visualizations
3. Analyze topic trends over time
4. Create interactive dashboards


```python
# Cell 1: Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted", font_scale=1.2)
```

```python
# Cell 2: Load dataset with topics
df = pd.read_csv("datasets/topics.csv")

print("✅ Dataset loaded:", df.shape)
print(df.head())
```

```python
# Cell 3: Publications per year
if "year" in df.columns:
    plt.figure(figsize=(10,5))
    sns.countplot(data=df, x="year", order=sorted(df["year"].dropna().unique()))
    plt.title("Number of Publications per Year")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Cell 4: Top authors
if "authors" in df.columns:
    authors = (
        df["authors"]
        .dropna()
        .str.split(";")
        .explode()
        .str.strip()
        .value_counts()
        .head(20)
    )

    plt.figure(figsize=(10,6))
    sns.barplot(y=authors.index, x=authors.values)
    plt.title("Top 20 Authors by Number of Publications")
    plt.xlabel("Publications")
    plt.ylabel("Author")
    plt.tight_layout()
    plt.show()
```

```python
# Cell 5: Top journals
if "journal" in df.columns:
    journals = (
        df["journal"]
        .dropna()
        .value_counts()
        .head(15)
    )

    plt.figure(figsize=(10,6))
    sns.barplot(y=journals.index, x=journals.values)
    plt.title("Top 15 Journals by Publications")
    plt.xlabel("Publications")
    plt.ylabel("Journal")
    plt.tight_layout()
    plt.show()
```

```python
# Cell 6: Topic distribution
if "topic" in df.columns:
    plt.figure(figsize=(10,6))
    sns.countplot(data=df, x="topic", order=df["topic"].value_counts().index)
    plt.title("Distribution of Topics")
    plt.xlabel("Topic ID")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()
```

```python
# Cell 7: Heatmap (year vs topic)
if "year" in df.columns and "topic" in df.columns:
    pivot = pd.crosstab(df["year"], df["topic"])
    plt.figure(figsize=(12,8))
    sns.heatmap(pivot, cmap="YlGnBu")
    plt.title("Publications by Year and Topic")
    plt.xlabel("Topic ID")
    plt.ylabel("Year")
    plt.tight_layout()
    plt.show()
```
